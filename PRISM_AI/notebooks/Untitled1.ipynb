```python
import joblib
import pandas as pd
import numpy as np


products = joblib.load(
    "models/products.pkl"
)


vectorizer = joblib.load(
    "models/tfidf_vectorizer.pkl"
)


similarity_matrix = joblib.load(
    "models/recommendation_similarity.pkl"
)


print("Recommendation Model Loaded ✅")
```

```python
def recommend_products(product_name, n=5):

    index = products[
        products["ProductName"] == product_name
    ].index[0]


    similarity_scores = list(
        enumerate(
            similarity_matrix[index]
        )
    )


    similarity_scores = sorted(
        similarity_scores,
        key=lambda x:x[1],
        reverse=True
    )


    recommended_indexes = []


    for i,score in similarity_scores[1:n+1]:
        recommended_indexes.append(i)


    return products.iloc[
        recommended_indexes
    ][
        [
            "ProductID",
            "ProductName",
            "Brand",
            "Category",
            "Price",
            "Rating"
        ]
    ]
```

```python
recommend_products(
    "AirPods Pro",
    5
)
```
