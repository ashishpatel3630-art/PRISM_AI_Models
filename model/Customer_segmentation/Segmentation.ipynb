```python


# Create Age Groups
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 30, 40, 50, 100], labels=['<30', '30-39', '40-49', '50+'])

# Create Satisfaction Categories
df['SatisfactionLevel'] = pd.cut(df['SatisfactionScore'], bins=[0, 4, 7, 10], labels=['Low', 'Medium', 'High'])

# Now you can use countplot!
sns.countplot(x='AgeGroup', hue='SatisfactionLevel', data=df)
plt.show()
```

```python
# Visualize relationships between key financial metrics
sns.pairplot(df[['Age', 'Income', 'TotalSpend', 'SatisfactionScore']])
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

for col in ["SatisfactionLevel","CustomerID"]:
    df[col] = encoder.fit_transform(df[col])

print("Categorical Converted Successfully ✅")
```

```python
num_columns=[
    'Income',
    'AverageOrderValue',
    'MonthlySpend',
    'SatisfactionScore',
    'SatisfactionLevel'
    ]
```

```python
for col in num_columns:
    df[col] = df[col].fillna(df[col].median())

print("Numerical Missing Values Handled ✅")
```
