```python
remove_cols =[ 
            "CustomerID",
            "RegistrationDate",
            "ChurnRisk"
             ]
```

```python
num_cols = df_model.select_dtypes(
    include=np.number
).columns
```

```python
for col in num_cols:
    df_model[col] = df_model[col].fillna(
        df_model[col].median()
    )
```
